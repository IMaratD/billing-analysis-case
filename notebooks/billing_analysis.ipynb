# Billing Analysis Notebook

This notebook contains exploratory analysis of billing-related entities and simulated anomaly detection scenarios.

Focus areas:
- partner hierarchy
- billing visibility
- subscription consistency
- anomaly distribution

```python
import matplotlib.pyplot as plt

users["partner_type"].value_counts().plot(kind="bar")

plt.title("Partner Type Distribution")
plt.xlabel("Partner Type")
plt.ylabel("Users Count")

plt.show()
```

```python
anomalies = pd.DataFrame({
    "anomaly_type": [
        "duplicate_invoice",
        "duplicate_invoice",
        "unauthorized_visibility",
        "billing_mismatch",
        "inactive_service_billed",
        "inactive_service_billed",
        "inactive_service_billed"
    ]
})

anomalies
```

```python
anomalies["anomaly_type"].value_counts().plot(kind="bar")

plt.title("Billing Anomalies Distribution")
plt.xlabel("Anomaly Type")
plt.ylabel("Count")

plt.show()
```

```python
anomaly_severity = pd.DataFrame({
    "anomaly_type": [
        "duplicate_invoice",
        "unauthorized_visibility",
        "billing_mismatch",
        "inactive_service_billed"
    ],
    "severity": [
        "medium",
        "critical",
        "high",
        "high"
    ]
})

anomaly_severity
```

```python
anomaly_severity["severity"].value_counts().plot(kind="bar")

plt.title("Anomaly Severity Distribution")
plt.xlabel("Severity")
plt.ylabel("Count")

plt.show()
```

```python
business_impact = pd.DataFrame({
    "issue": [
        "duplicate_invoice",
        "unauthorized_visibility",
        "billing_mismatch",
        "inactive_service_billed"
    ],
    "business_impact": [
        "financial_risk",
        "security_risk",
        "revenue_loss",
        "customer_dissatisfaction"
    ]
})

business_impact
```

```python
business_impact["business_impact"].value_counts().plot(kind="bar")

plt.title("Business Impact Distribution")
plt.xlabel("Impact Type")
plt.ylabel("Count")

plt.show()
```

# Conclusions

The analysis demonstrates several important aspects of billing system complexity in SaaS products:

- partner hierarchy significantly affects billing behavior;
- inherited billing configuration increases logical complexity;
- access control inconsistencies may lead to security risks;
- billing anomalies may directly affect revenue and customer trust;
- analytical monitoring can help detect inconsistencies before they affect production systems.

The investigation also highlights the importance of:
- anomaly detection;
- business-oriented analysis;
- validation of billing logic;
- monitoring of subscription-related behavior.
